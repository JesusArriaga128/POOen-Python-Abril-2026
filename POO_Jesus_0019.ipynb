#### Conceptos de POO (Programación Orientada a Objetos) en Python

## El promt en IA gemini
Lenguaje Python, crear la clase "animal" con los atributos "nombre del animal", "raza" crear el constructor animal, y las funciones ladrar (), correr(), crear los objetos pitbull, chihuahua, nivel principiante no es interactivo, proporciona tu los datos

```python
# Definición de la clase
class Animal:
    # El constructor __init__ inicializa los atributos del objeto
    def __init__(self, nombre_del_animal, raza):
        self.nombre_del_animal = nombre_del_animal
        self.raza = raza

    # Función para que el animal ladre
    def ladrar(self):
        print(f"{self.nombre_del_animal} dice: ¡Guau guau!😢")

    # Función para que el animal corra
    def correr(self):
        print(f"{self.nombre_del_animal} está corriendo velozmente.✨")
    def morder(self):
        print(f"{self.nombre_del_animal} está mordiendo a la vecina 🤷‍♀️")

# --- Creación de los objetos (Instancias) ---

# Objeto 1: Pitbull
pitbull = Animal("Rocky", "Pitbull")

# Objeto 2: Chihuahua
chihuahua = Animal("Chispa", "Chihuahua")
# Creamos el objeto perro callejero con datos
perrocallejero = Animal("Solovino","Chafa")

# --- Uso de los objetos ---

print("--- Datos del Pitbull ---")
print(f"Nombre: {pitbull.nombre_del_animal}")
print(f"Raza: {pitbull.raza}")
pitbull.ladrar()
pitbull.correr()

print("\n--- Datos del Chihuahua ---")
print(f"Nombre: {chihuahua.nombre_del_animal}")
print(f"Raza: {chihuahua.raza}")
chihuahua.ladrar()
chihuahua.correr()
chihuahua.morder()

print("\n--- Datos del perro callejero ---")
print(f"Nombre: {perrocallejero.nombre_del_animal}")
print(f"Raza: {perrocallejero.raza}")
perrocallejero.ladrar()
perrocallejero.correr()
perrocallejero.morder()

print("Programa realizado por Jesus Arriaga NL 5")
```
